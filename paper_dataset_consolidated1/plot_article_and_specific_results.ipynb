## Modelo e parâmetros comuns

No quadro de interação e em ressonância, o Hamiltoniano é

$$\hat H(t)=\lambda(t)\left(\hat\sigma_+\hat a+\hat\sigma_-\hat a^\dagger\right).$$

São utilizados dois perfis de acoplamento:

$$\lambda_{\mathrm c}(t)=\lambda_0\cos(\omega t+\phi),$$

$$\lambda_{\mathrm G}(t)=\lambda_0\exp\!\left[\frac{\vartheta(t-T)^2}{\zeta^2}\right].$$

Em todas as contas desta primeira parte: $\lambda_0=1$, $\vartheta=-1$ no perfil gaussiano, $\phi=0$ no perfil cossenoidal, $\alpha=\sqrt{5}$, $\langle\hat n\rangle=|\alpha|^2=5$, dimensão atômica $N=2$ e truncamento do campo $N_b=45$.

A dinâmica aberta usa $\kappa=10^{-1}$, $\gamma_\phi=10^{-2}$ e $\gamma=0$. Os operadores de colapso implementados são

$$\hat c_\kappa=\sqrt{\kappa}\,\hat a,\qquad \hat c_\phi=\sqrt{\gamma_\phi}\,\hat\sigma_z.$$

A malha temporal é mais fina entre $0$ e $1$. Os mapas densos usam as coordenadas físicas dessa malha não uniforme; portanto, a maior quantidade de amostras iniciais não alarga artificialmente essa região.

## 1. Negatividade da função de Wigner do campo

O estado inicial é um gato par no campo, desacoplado do estado excitado do átomo:

$$|\psi_W(0)\rangle=\mathcal N_+|e\rangle\left(|\alpha\rangle+|-\alpha\rangle\right),$$

$$\mathcal N_+=\left[2\left(1+e^{-2|\alpha|^2}\right)\right]^{-1/2}.$$

O quantificador é

$$\mathcal N_W(\rho_F)=\frac12\left[\iint |W(q,p)|\,dq\,dp-1\right].$$

Os gráficos mostram a negatividade de Wigner normalizada. Valor positivo indica volume negativo da função de Wigner e, portanto, não classicalidade do campo. O tempo é $0\le t\le15$.

## 2. Coerência atômica

O átomo começa no estado maximamente coerente $|+\rangle$ e o campo em um estado coerente:

$$|\psi_C(0)\rangle=\frac{|e\rangle+|g\rangle}{\sqrt2}\otimes|\alpha\rangle.$$

A quantidade correta usada nos arquivos e nos gráficos é a entropia relativa de coerência da matriz reduzida do átomo, com logaritmo em base 2:

$$\mathcal C_{\mathrm{rel}}(\rho_Q)=S(\rho_{Q,\mathrm{diag}})-S(\rho_Q),\qquad \rho_Q=\operatorname{Tr}_F\rho.$$

Logo, $0\le\mathcal C_{\mathrm{rel}}\le1$ para o qubit. O tempo é $0\le t\le50$.

## 3. Magia atômica de estado misto

O estado inicial é

$$|\psi_M(0)\rangle=\left[\cos\left(\frac\theta2\right)|e\rangle+e^{i\pi/4}\sin\left(\frac\theta2\right)|g\rangle\right]\otimes|\alpha\rangle,$$

com $\cos\theta=1/\sqrt3$. Como a evolução é aberta, o antigo $M_2$ não é usado como certificado de magia. Os gráficos usam o testemunho misto

$$W_{1/2}(\rho_Q)=2\ln\left[\frac{1+|\langle\sigma_x\rangle|+|\langle\sigma_y\rangle|+|\langle\sigma_z\rangle|}{2}\right].$$

Para um único qubit, $W_{1/2}>0$ se e somente se $|\langle\sigma_x\rangle|+|\langle\sigma_y\rangle|+|\langle\sigma_z\rangle|>1$, isto é, o estado reduzido está fora do octaedro estabilizador. $W_{1/2}$ é um testemunho, não um monotônico de recurso. O tempo é $0\le t\le50$.

As três varreduras são $\zeta\in[6,10]$ com $T=25$, $T\in[15,35]$ com $\zeta=8$, e $\omega\in[0,\pi/5]$ com $\phi=0$. Os pontos destacados são os mesmos da subseção de coerência.

## 4. Emaranhamento átomo-campo

O estado inicial correlaciona o átomo com dois estados coerentes opostos do campo:

$$|\psi_E(0)\rangle=\frac{1}{\sqrt2}\left(|e\rangle|\alpha\rangle+|g\rangle|-\alpha\rangle\right).$$

A normalização é exatamente $1/\sqrt2$ porque $\langle e|g\rangle=0$. Para $\alpha=\sqrt5$, esse estado apresenta negatividade logarítmica inicial muito próxima de 1.

O quantificador é a negatividade logarítmica

$$E_{\mathcal N}(\rho)=\log_2\left\|\rho^{T_Q}\right\|_1.$$

$E_{\mathcal N}>0$ sinaliza emaranhamento detectado entre átomo e campo. O tempo é $0\le t\le50$. As varreduras e os pontos destacados são os mesmos da coerência e da magia: $\zeta\in[6,10]$, $T\in[15,35]$ e $\omega\in[0,\pi/5]$.
